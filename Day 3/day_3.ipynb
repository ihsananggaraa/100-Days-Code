## **If Else**

* Secara sederhana if : Jika, else: Maka
* Menyatakan suatu kondisi
* Contoh :

    Jika saya lapar saya makan, jika tidak lapar saya tidur. Maka dapat ditulis di kode secara sederhananya sebagai berikut:

    ```python
    if saya lapar:
        saya makan
    else:
        saya tidur
    ```
