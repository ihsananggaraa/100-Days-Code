* int : Integer (bilangan bulat)
* float : bilangan pecahan / desimal

# **Operasi Matematika**

* (*) Untuk perkalian
* (+) Untuk penjumlahan
* (%) Untuk modulo (sisa)
* (/) Untuk pembagian
* (-) Untuk Pengurangan
* (**) Untuk pangkat

## **Contoh Implementasi Operasi Matematika**

```python
print("Ini adalah hasil perkalian 5 dan 5: ", 5*5)
print("Ini adalah hasil pengurangan 10 dan 5: ", 10-5)

print("Ini adalah hasil perkalian 7 dan 2: ", 7*2)
print("Ini adalah hasil pembagian 10 dan 2: ", 10/2)

print("Ini adalah hasil pemangkatan 10 pangkat 2: ", 10**2)
print("Ini adalah hasil sisa bagi 100 dan 2: ", 100%2)
```

## - **Pizza Restaurant**

```python
welcome_message ="Wellcome to the pizza restaurant"
print("=" * len(welcome_message),"\n"
      ,welcome_message)

print("="*len(welcome_message))
print(welcome_message)

# #Pizza category
# pizza_size = ['Small', 'Medium','Large']
# pizza_topping = ['Tomato', 'Cheese', 'Beef']
# additional_topping = ['Mozarela', 'Pepperoni']


pizza_size = (input("Please choose pizza size (Small, Medium, Large)\n"))
price_pizza = 0
if pizza_size == 'Small':
    price_pizza +=1
elif pizza_size == 'Medium':
    price_pizza +=3
elif pizza_size == 'Large':
    price_pizza +=5
else:
    print("Please input answer correctly")
print(price_pizza)

pizza_topping= (input("Please choose pizza topping (Tomato, Cheese, Beef\n"))
price_topping = 0
if pizza_topping == 'Tomato':
    price_topping +=1 #ingat bahwa =+b dengan +=b itu beda
elif pizza_topping == 'Cheese':
    price_topping +=3
elif pizza_topping == 'Beef':
    price_topping +=5
else:
    print("Please input answer correctly")

add_topping= input("Please choose additional topping (Mozarela, Pepperoni)\n")
price_add = 0
if add_topping == 'Mozarela':
    price_topping +=1 #ingat bahwa =+b dengan +=b itu beda
elif pizza_topping == 'Pepperoni':
    price_topping +=3
else:
    print("Please input answer correctly")
print(price_add)

price_total = price_pizza + price_topping + price_add
print(f"Your Total Price was {price_total}$")

bill= input("Do you want add bills? Y/N \n") #gaperlu print
bill_total = 0
if bill == "Y":
    paid_bill = int(input("How much bill you want to paid?"))
    person_bill = int(input("How many person would pay for bill"))
    bill_per_person = paid_bill / person_bill
    if person_bill <= 0 :
        print("Please input the answer correctly")
    elif person_bill == 1:
        bill_total += bill_per_person
        print(f"You must paid for{bill_total}$")
    else:
        bill_total += bill_per_person
        print(f"Each person must pay for{bill_per_person:.2f}$")
else:
    print("Thank you for coming")
print(f"Your expend today is {price_total}$ for pizza and {bill_total:.2f}$ for bill")
print("="* len("Thank you for coming"))
```

## - **Kalkulator BMI**

```python
print("Ini adalah program untuk menghitung BMI")

name = input("Masukkan nama anda: ")
born = int(input("Masukkan tahun lahir anda: "))
age = int(2025-born)
weight = int(input("Masukkan berat badan anda: "))
height = int(input("Masukkan tinggi badan anda: "))
bmi = round(weight/((height/100)**2),2)
lower = 18.5
if bmi < lower:
    category = "kurus"
elif bmi >= 18.5 and bmi <= 24.9:
    category = "ideal"
else: category = "over weight"
print(f"Halo {name}, kamu lahir di tahun {born} ya, Umur kamu sekaramg {age} tahun."
      f"BMI kamu berada di angka {bmi}, dengan kategori {category}.")
```
