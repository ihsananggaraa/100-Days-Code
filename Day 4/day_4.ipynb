**List**


Modul `random` bawaan Python digunakan untuk menghasilkan angka acak semu (pseudo-random) yang didasarkan pada algoritma matematika (Mersenne Twister) untuk berbagai keperluan komputasi dan simulasi.

**Fungsi Pembangkit Angka Dasar**

* random(): Menghasilkan angka desimal (float) acak antara 0.0 hingga 1.0 (tidak termasuk 1.0).

* randint(a, b): Menghasilkan bilangan bulat (integer) acak antara a dan b (termasuk batas a dan b).

* randrange(start, stop, step): Menghasilkan bilangan bulat acak dari rentang tertentu dengan kelipatan langkah (step) tertentu, persis seperti fungsi range() bawaan Python.

* uniform(a, b): Menghasilkan angka desimal acak yang berdistribusi seragam (rata) di antara rentang a dan b.

**Fungsi Manipulasi Urutan (List/Sequence)**

* choice(seq): Mengambil satu elemen secara acak dari tipe data berurut (list, tuple, atau string).

* choices(population, weights=None, k=1): Memilih k jumlah elemen secara acak dengan pengembalian (with replacement). Anda bisa memberikan parameter weights untuk menentukan probabilitas kemunculan tiap elemen.

* sample(population, k): Mengambil k elemen unik secara acak tanpa pengembalian (without replacement), sangat berguna untuk pengambilan sampel data.

* shuffle(x): Mengacak susunan elemen di dalam sebuah list in-place (mengubah urutan pada list aslinya secara permanen).

**Fungsi Distribusi Statistik Khusus**

* gauss(mu, sigma) / normalvariate(mu, sigma): Menghasilkan angka acak yang mengikuti distribusi Normal (Gaussian) dengan nilai rata-rata mu dan simpangan baku sigma.

* expovariate(lambd): Menghasilkan angka acak berdasarkan distribusi eksponensial. Modul ini juga mendukung distribusi lain seperti betavariate(), gammavariate(), dan paretovariate().

**Pengendalian Generator (Seed)**

* seed(a): Menetapkan nilai awal untuk algoritma pengacak. Menjalankan kode dengan nilai seed yang sama persis akan selalu menghasilkan urutan keluaran angka acak yang berulang secara identik. Fungsi ini wajib digunakan ketika Anda membutuhkan hasil yang dapat direproduksi (reproducible) saat pengujian atau pemodelan data.

# **Latihan**

```python
kota_indo = ["Surabaya", "Jakarta", "Medan"]
print(kota_indo[0]) #bukan print([0]), like "Print list kota_indo urutan ke 1"
# Ingat bahwa: programmers start counting from zero

print(kota_indo[-1]) # dari urutan belakang

kota_indo[2] = "Bandung"  #Mengganti indeks ke 3 "Medan" Menjadi Bandung
print(kota_indo)

kota_indo.append("Malang") #menambahkan setelah terakhir
print(kota_indo)

kota_indo.extend("Jawa")
print(kota_indo)

kota_indo.extend(["Batu", "Kediri"])
print(kota_indo)
```

```python
rock = '''
    _______
---'   ____)
      (_____)
      (_____)
      (____)
---.__(___)
'''

paper = '''
    _______
---'   ____)____
          ______)
          _______)
         _______)
---.__________)
'''

scissors = '''
    _______
---'   ____)____
          ______)
       __________)
      (____)
---.__(___)
'''

import random
welcome_message = "Wellcome to the Paper, Rock, Scissors game"
print("=" * len(welcome_message))
print(welcome_message)

prs = ["rock", "paper", "scissors" ]
computer= prs[random.randint(0,2)]
user_input = int(input("Masukkan pilihan anda (0 : Rock , 1: Paper, 2: Scisssors)\n"))
if user_input == 0 and computer ==prs[1]:
    print("Your choice:\n" + rock + "\nvs\n" +"computer choice"+ paper)
    print("Computer Win")
elif user_input == 0 and computer ==prs[2]:
    print("Your choice:\n" + rock + "\nvs\n" + "computer choice" + scissors)
    print("You are win")
elif user_input == 0 and computer ==prs[user_input]:
    print("Your choice:\n" + rock + "\nvs\n" + "computer choice" + rock)
    print("Draw")
elif user_input == 1 and computer ==prs[user_input]:
    print("Your choice:\n" + paper + "\nvs\n" + "computer choice" + paper)
    print("Draw")
elif user_input == 1 and computer ==prs[0]:
    print("Your choice:\n" + paper + "\nvs\n" + "computer choice" + rock)
    print("You are win")
elif user_input == 1 and computer ==prs[2]:
    print("Your choice:\n" + paper + "\nvs\n" + "computer choice" + scissors)
    print("Computer are win")
elif user_input == 2 and computer ==prs[user_input]:
    print("Your choice:\n" + scissors + "\nvs\n" + "computer choice" + scissors)
    print("Draw")
elif user_input == 2 and computer ==prs[0]:
    print("Your choice:\n" + scissors + "\nvs\n" + "computer choice" + rock)
    print("Computer are win")
elif user_input == 2 and computer ==prs[1]:
    print("Your choice:\n" + scissors + "\nvs\n" + "computer choice" + paper)
    print("You are win")
else:
    print("Masukkan angka dengan benar")
```
